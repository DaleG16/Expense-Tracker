```python
import numpy as np
import json
import os
if os.path.exists('expenses.json'):
    with open('expenses.json', 'r') as file:
        expenses=json.load(file)
else:
    expenses=[]

while True:
  print("1.Add Expense")
  print("2.View Expenses")
  print("3.Total Spending")
  print("4.Spending by Category")
  print("5. EXIT")
  try:
    choice=int(input("Enter your choice: "))
  except ValueError:
    print("Enter only the numbers; not letters. TRY AGAIN!\n")
    continue
  if choice==1:
    expense={}
    name=input("Enter the expense name: ")
    while True:
      try:
        amount=int(np.abs(int(input("Enter the amount: ")))) #numpy is int64 which is different from python int so 
          #The outer int() converts the NumPy result back into a standard Python integer.
        break
      except ValueError:
        print("\nPlease enter only the numbers; not letters. Try again!")
    category=input("Enter the category(ex: Food,Transport): ")



    expense={'name':name,'amount':amount,'category':category}
    expenses.append(expense)
    with open('expenses.json', 'w') as file:
        json.dump(expenses,file,indent=4)
    print("Added Expense Successfully\n")
  elif choice==2:
    if not expenses:
      print("No expenses recored yet\n")
    else:
      print("\n---Your Expenses---")
      for item in expenses: #Since each dictionary now has fixed keys, you no longer need a nested loop:
        print(f"{item['name']}:₹{item['amount']} | Category:{item['category']}")
      print("\n")
        
  elif choice==3:
    total=0#Every calculation startys fresh
    if not expenses:
      print("No expenses recorded yet!. Please add the expenses\n")
    else:
      for item in expenses: #calculate the amount directly
        total+=item['amount']
      print("Total Expenses: ",total,"\n")

  elif choice==4:
    print("\n ---Expenses based on Category---")
    category_name=input("Enter the Category name: ")
    total=0
    found=False
    for item in expenses:
      
      if item['category'].lower()==category_name.lower():
        total+=item['amount']
        found=True
    if found:
      print(f"Category: {category_name}")
      print(f"Total Spending: ₹{total}\n")
    else:
      print("No expenses found for this category")

      
  elif choice==5:
    print("Program Exiting...")
    break
  else:
    print("Invalid choice. TRY AGAIN!\n")
    
print(expenses)
```
